```python
from Bio import SeqIO
import random


def sample_fasta(
    input_fasta: str,
    output_fasta: str,
    n: int = 200,
    seed: int = 42,
):
    """
    Randomly sample n sequences from a FASTA file and write them
    to a new FASTA file.

    Sampling is reproducible given the same seed.
    """
    records = list(SeqIO.parse(input_fasta, "fasta"))

    if n > len(records):
        raise ValueError(
            f"Requested {n} sequences, but FASTA only contains {len(records)}."
        )

    rng = random.Random(seed)
    sampled_records = rng.sample(records, n)

    SeqIO.write(sampled_records, output_fasta, "fasta")

    print(
        f"Sampled {n} / {len(records)} sequences "
        f"and wrote them to {output_fasta}"
    )

sample_fasta(
    input_fasta="/home/akhanna/AMP/SILO_amp_new/objectives/reference_data/antibacterial.fasta",
    output_fasta="/home/akhanna/AMP/SILO_amp_new/objectives/reference_data/antibacterial_ref.fasta",
    n=200,
    seed=42,
)

sample_fasta(
    input_fasta="/home/akhanna/AMP/SILO_amp_new/training/training_data/final_train.fasta",
    output_fasta="/home/akhanna/AMP/SILO_amp_new/objectives/reference_data/training_ref.fasta",
    n=200,
    seed=42,
)
```
